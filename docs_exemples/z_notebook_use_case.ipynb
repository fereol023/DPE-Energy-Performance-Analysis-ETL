```python
from dpe_enedis_ademe_etl_engine.pipelines import DataEnedisAdemeExtractor

pipeline = DataEnedisAdemeExtractor()

os.environ['PREFECT_API_URL'] = ''
pipeline.extract.fn(
    pipeline,
    from_input=False, 
    code_departement=78,  
    annee=2023, 
    rows=10, 
    n_threads_for_querying=5
    )
```

```python
pipeline.output.batch_id.value_counts()
```

```python
# save above results into a db - files are already saved in the path set in env vars
import pandas as pd
from sqlalchemy import create_engine

uri = f"postgresql://{os.environ['POSTGRES_ADMIN_USERNAME']}:{os.environ['POSTGRES_ADMIN_PASSWORD']}@{os.environ['POSTGRES_HOST']}:{os.environ['POSTGRES_PORT']}/{os.environ['POSTGRES_DB_NAME']}"
obj_engine = create_engine(uri)
pipeline.output.to_sql(
    "raw_extract",
    con=obj_engine,
    if_exists="append",
    index=False
)

pd.read_sql("select * from raw_extract limit 10", con=obj_engine)
```

```python
# transfor and load steps

from dpe_enedis_ademe_etl_engine.pipelines import DataEnedisAdemeTransformer, DataEnedisAdemeLoader


pipeline_transform = DataEnedisAdemeTransformer(
    df=pipeline.output,
    golden_data_config_fpath="z_golden_schema.json"
)
pipeline_transform.run()
```

```python
pipeline_load = DataEnedisAdemeLoader(
    engine=obj_engine,
    debug=True)
pipeline_load.run()
```
